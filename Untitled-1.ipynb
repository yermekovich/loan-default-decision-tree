```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt

# Загружаем большой датасет
url = "https://raw.githubusercontent.com/selva86/datasets/master/GermanCredit.csv"
df = pd.read_csv(url)

# Выбираем признаки, похожие на пример (числовые, чтобы не усложнять кодирование)
# amount ≈ Income, age, duration, installment_rate
features = ['amount', 'age', 'duration', 'installment_rate']
X = df[features]
y = df['credit_risk']  # 1 - одобрено (good risk), 0 - отказ (bad risk)

print("Размер датасета:", df.shape)
print("Первые 5 строк:\n", df.head())

# Разделение на train/test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Инициализация модели (gini, max_depth=3 как в примере)
clf = DecisionTreeClassifier(criterion='gini', max_depth=3, random_state=42)

# Обучение
clf.fit(X_train, y_train)
```

```python
plt.figure(figsize=(12, 8))
plot_tree(clf, 
          feature_names=features, 
          class_names=['Reject (bad)', 'Approve (good)'], 
          filled=True, 
          rounded=True)
plt.title("Дерево решений (German Credit Dataset, criterion=gini)")
plt.show()
```

```python
y_pred = clf.predict(X_test)

print("Матрица ошибок (Confusion Matrix):")
print(confusion_matrix(y_test, y_pred))

print("\nОтчет о классификации:")
print(classification_report(y_test, y_pred))
```

```python
# Добавляем 5 противоречивых строк
new_data = pd.DataFrame({
    'amount': [8000, 12000, 15000, 9000, 11000],
    'age': [35, 28, 45, 30, 40],
    'duration': [48, 60, 36, 24, 48],
    'installment_rate': [4, 3, 5, 2, 4],
    'credit_risk': [0, 0, 0, 0, 0]  # deliberately bad risk
})

df_extended = pd.concat([df, new_data], ignore_index=True)

# Переобучаем на расширенном датасете
X_ext = df_extended[features]
y_ext = df_extended['credit_risk']
X_train_ext, X_test_ext, y_train_ext, y_test_ext = train_test_split(X_ext, y_ext, test_size=0.3, random_state=42)

clf_ext = DecisionTreeClassifier(criterion='gini', max_depth=3, random_state=42)
clf_ext.fit(X_train_ext, y_train_ext)

y_pred_ext = clf_ext.predict(X_test_ext)
print("Отчет после добавления противоречий:")
print(classification_report(y_test_ext, y_pred_ext))
```
